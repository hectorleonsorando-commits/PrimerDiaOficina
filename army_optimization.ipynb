# Army Optimization using Linear Programming

This notebook demonstrates how to use linear programming to optimize army recruitment by maximizing power given resource constraints.

## Import Required Libraries

Import necessary libraries such as numpy and scipy.optimize for linear programming.

```python
import numpy as np
from scipy.optimize import linprog
```

## Define the Linear Programming Problem

Define the decision variables for the number of swordsmen, bowmen, and horsemen.

## Set Up Constraints

Specify the constraints based on available food, wood, and gold resources.

```python
# Coefficients for the objective function (maximize power, so minimize negative)
c = [-70, -95, -230]

# Coefficients for the inequality constraints (A_ub @ x <= b_ub)
A_ub = [
    [60, 80, 140],  # Food constraint
    [20, 10, 0],    # Wood constraint
    [0, 40, 100]    # Gold constraint
]

b_ub = [1200, 800, 600]

# Bounds for variables (non-negative)
bounds = [(0, None), (0, None), (0, None)]
```

## Define the Objective Function

Set the objective to maximize the total power of the army.

## Solve the Optimization Problem

Use scipy.optimize.linprog to solve the linear programming problem.

```python
# Solve the linear programming problem
result = linprog(c, A_ub=A_ub, b_ub=b_ub, bounds=bounds, method='highs')

# Check if the optimization was successful
if result.success:
    swordsmen, bowmen, horsemen = result.x
    max_power = -result.fun  # Since we minimized -power, negate to get max power
else:
    print("Optimization failed:", result.message)
```

## Display the Results

Print the optimal number of each unit and the maximum power achieved.

```python
print(f"Optimal number of Swordsmen: {swordsmen:.0f}")
print(f"Optimal number of Bowmen: {bowmen:.0f}")
print(f"Optimal number of Horsemen: {horsemen:.0f}")
print(f"Maximum Power: {max_power:.0f}")
```
